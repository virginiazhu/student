# Task 1 Simple Loot Box

```python
import random
p = float(input("Rare chance % (10–90): ") or 30)/100
common = rare = 0
for _ in range(10):
    rare += random.random() < p
common = 10 - rare
print(f"Rare: {rare}\nCommon: {common}")
```

# Task 2 Dice Rolling Counter

```python
import random
cnt = [0]*6
for _ in range(20):
    cnt[random.randint(1,6)-1] += 1
for i,c in enumerate(cnt,1):
    print(i, c)
```

# Task 3 Random Story Generator

```python
import random

who = ["knight", "wizard", "robot", "pirate", "alien"]
where = ["castle", "forest", "space station", "island", "mountain"]
act = ["found treasure", "built a time machine", "fought monsters", "discovered magic", "took a nap"]

pick = lambda a: random.choice(a)

print("🌟 Random Adventure Story 🌟\n")
for _ in range(5):
    print(f"One day, the {pick(who)} traveled to the {pick(where)} and {pick(act)}.")
```

# Task 4 Basic Fairess Test

```python
import random

rolls = [random.randint(1, 6) for _ in range(100)]

counts = {i: rolls.count(i) for i in range(1, 7)}

print("Roll counts:")
for number, count in counts.items():
    print(f"{number}: {count}")

# Check fairness: each should be close to 16-17
fair = all(14 <= count <= 19 for count in counts.values())  # Allow a small margin

if fair:
    print("Fair!")
else:
    print("Maybe unfair?")
```
