<h2>Sprint 2: Javascript/Python Lessons</h2>
## CSP Big Idea 3 - Sprint 2 Key Takeaways

**3.1 Variables**  
I learned that variables store, assign, and update data, acting as named containers for information. I learned that using descriptive names improves readability and helps track changing data throughout a program.  

**3.2 Data Abstraction**  
I learned that data abstraction simplifies complex information into manageable forms. I learned that lists allow multiple values to be stored under one name, making it easier to manipulate, iterate, and modify related data efficiently.  

**3.3 Mathematical Expressions**  
I learned that mathematical expressions combine numbers, variables, and operators to produce new values. I learned that using parentheses and understanding operator precedence ensures calculations occur in the correct order and prevents logical errors. 

**3.4 Strings**  
I learned that strings represent sequences of characters as a single value. I learned that treating text as a single data type allows programs to manipulate, analyze, and display textual information effectively.  

**3.6 Conditionals**  
I learned that conditionals allow programs to make decisions by evaluating expressions. I learned that

 comparison and logical operators help direct program flow through `if` and `else` statements.  

**3.7 Nested Conditionals**  
I learned that nested conditionals enable multiple layers of decision-making within a program. I learned that they allow programs to evaluate complex scenarios that depend on several conditions simultaneously. 

```python
if age >= 18:
    if has_id:
        print("Entry allowed") 
```

**3.8 Iterations**  
I learned that iteration structures like `for` and `while` loops let programs repeat actions automatically. 

I learned that loops help reduce code repetition and allow programs to process sequences or continue operations while conditions are true.  

**3.9 Developing Algorithms**  
I learned that algorithms are step-by-step procedures for handling inputs, performing logical operations, and producing outputs. I learned that designing algorithms before coding helps break problems into manageable steps and ensures program correctness.  

**3.10 Lists**  
I learned that lists store multiple values under a single variable name. I learned that they allow indexing, modification, and iteration, making it easier to organize and track data that changes over time.  

```python
# A list storing multiple values
fruits = ["apple", "banana", "cherry"]

# Accessing an element by index
print(fruits[0])  # Output: apple

# Adding a new element
fruits.append("orange")

# Iterating through the list
for fruit in fruits:
    print(fruit)
```

**3.12 Calling Procedures**  
I learned that calling procedures executes prewritten blocks of code. I learned that using procedures improves program organization, reduces repetition, and makes code easier to read and maintain.  

**3.13 Developing Procedures**  
I learned that procedures are reusable blocks of code defined with parameters. I learned that creating procedures helps organize complex programs, supports testing and debugging, and allows code to be reused without rewriting it.  

**3.15 Random Values**  
I learned that random values introduce unpredictability in programs. I learned that using randomness can simulate real-world scenarios and handle varied inputs and outputs.  

**3.17 Alogrithim Efficiency**  
I learned that writing efficient code minimizes unnecessary computation and memory use. I learned the big O notation and how exponential and factorial algorithms are considered unefficent
