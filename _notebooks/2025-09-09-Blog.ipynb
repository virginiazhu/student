## 🎨 Backgrounds 

We also explored **backgrounds** as part of an adventure game project:  

### 🖼️ What Are Backgrounds?  
Backgrounds are visual elements that fill the space behind the main content or objects in a game or web application. They:  
- Provide **context**  
- Create **atmosphere**  
- Improve **visual appeal**  

### 🔑 Key Properties  
- **Background Color**: solid color behind content  
- **Background Image**: can be fixed, repeated, or scaled  
- **Positioning**: where the background starts  
- **Size & Repeat**: controls stretching or tiling

````python
import tkinter as tk
from tkinter import messagebox, simpledialog

# ------------------ Game Data ------------------ #
JEOPARDY_DATA = {
    "Boolean Basics": [
        ("Boolean values can only be ____ or ____.", "True or False", 100),
        ("What type does a Boolean expression return?", "Boolean", 200),
        ("What is the Boolean value of 0 in Python?", "False", 300),
        ("What is the Boolean value of 1 in Python?", "True", 400),
        ("Expression with comparison operators are called?", "Boolean expressions", 500)
    ],
    "Relational Operators": [
        ("What does 5 == 5 evaluate to?", "True", 100),
        ("What does 10 >= 10 evaluate to?", "True", 200),
        ("What does 7 < 3 evaluate to?", "False", 300),
        ("What does 12 != 12 evaluate to?", "False", 400),
        ("What does 8 <= 10 evaluate to?", "True", 500)
    ],
    "Logical Operators": [
        ("True and False = ?", "False", 100),
        ("False or True = ?", "True", 200),
        ("not(False or True) = ?", "False", 300),
        ("True and not False = ?", "True", 400),
        ("(True or False) and False = ?", "False", 500)
    ],
    "Evaluate It!": [
        ("(3 < 5) and (10 > 1) = ?", "True", 100),
        ("(7 == 8) or (5 < 2) = ?", "False", 200),
        ("not(6 <= 6) = ?", "False", 300),
        ("(4 != 5) and (2 > 8) = ?", "False", 400),
        ("(10 >= 10) or (3 == 4) = ?", "True", 500)
    ],
    "Mixed Review": [
        ("What keyword in Python negates a condition?", "not", 100),
        ("Is 0 considered True or False?", "False", 200),
        ("Is -1 considered True or False?", "True", 300),
        ("What is the opposite of and?", "or", 400),
        ("What is the opposite of or?", "and (with negation)", 500)
    ]
}

# ------------------ Game Class ------------------ #
class JeopardyGUI:
    def __init__(self, root):
        self.root = root
        self.root.title("Boolean Expressions Jeopardy")
        self.score = 0
        self.buttons = {}  
        self.flipped = {}  

        # Scoreboard
        self.score_label = tk.Label(root, text=f"Score: {self.score}", font=("Arial", 14))
        self.score_label.grid(row=0, column=0, columnspan=len(JEOPARDY_DATA), pady=10)

        # Categories
        for col, cat in enumerate(JEOPARDY_DATA.keys()):
            lbl = tk.Label(root, text=cat, font=("Arial", 12, "bold"))
            lbl.grid(row=1, column=col, padx=10, pady=5)

        # Question buttons
        for col, (cat, qlist) in enumerate(JEOPARDY_DATA.items()):
            for row_index, (q, a, val) in enumerate(qlist):
                btn = tk.Button(root,
                                text=f"${val}",
                                width=20, height=3,
                                command=lambda c=cat, i=row_index: self.flip_card(c, i))
                btn.grid(row=2 + row_index, column=col, padx=5, pady=5)
                self.buttons[(cat, row_index)] = btn

    def flip_card(self, category, idx):
        btn = self.buttons[(category, idx)]
        q, a, val = JEOPARDY_DATA[category][idx]
        state = self.flipped.get((category, idx), 0)

        if state == 0:  # Show question
            btn.config(text=q, wraplength=150, justify="center")
            self.flipped[(category, idx)] = 1

        elif state == 1:  # Show answer + ask correctness
            btn.config(text=f"Answer: {a}", wraplength=150, justify="center")
            self.flipped[(category, idx)] = 2

            # Ask user if correct
            correct = messagebox.askyesno("Did you get it right?", f"Answer: {a}\n\nDid you answer correctly?")
            if correct:
                self.add_score(val)
            else:
                self.add_score(-val)

            # Disable button after use
            btn.config(state="disabled")

    def add_score(self, points):
        self.score += points
        self.score_label.config(text=f"Score: {self.score}")


# ------------------ Run Game ------------------ #
def main():
    root = tk.Tk()
    game = JeopardyGUI(root)
    root.mainloop()

if __name__ == "__main__":
    main()
````
