### Boolean basics
- Values: `True` or `False`. Used for decisions and flow control.

### Relational operators

| Operator | Meaning        | Example     | Result |
|---------:|----------------|-------------|-------:|

| `>=`     | ≥              | `6 >= 6`    | True  |
| `<=`     | ≤              | `9 <= 4`    | False |

### Modulus operator `%`
- `num % 2 == 0` → even check
- `num % 2 == 1` → odd check
- `day % 7` → cycle through weekdays

### Logical operators
- `not X` — negation (flips True/False)
- `X and Y` — True only if both are True
- `X or Y` — True if at least one is True
